```python

import matplotlib.pyplot as plt

time_windows = [
    "60 Seconds",
    "1 Hour",
    "24 Hours"
]

matching_counts = [36, 639, 679]
total_duplicates = 679

# Calculate matching percentages
matching_rates = [
    count / total_duplicates * 100
    for count in matching_counts
]

# Create figure
fig, ax = plt.subplots(figsize=(11, 6), dpi=150)

colors = ["#94A3B8", "#2563EB", "#16A34A"]

bars = ax.bar(
    time_windows,
    matching_rates,
    color=colors,
    width=0.55
)

# Add exact values above each bar
for bar, rate, count in zip(
    bars,
    matching_rates,
    matching_counts
):
    ax.annotate(
        f"{rate:.1f}%\n({count:,}/{total_duplicates:,})",
        xy=(
            bar.get_x() + bar.get_width() / 2,
            rate
        ),
        xytext=(0, 9),
        textcoords="offset points",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

# Chart title
ax.set_title(
    "One-Hour Matching Identifies 94.1% of\n"
    "Retrospective Duplicate-Payment Labels",
    fontsize=15,
    fontweight="bold",
    pad=24
)

ax.set_ylabel(
    "Duplicate-Labeled Payments With a Match (%)",
    fontsize=11
)

ax.set_xlabel(
    "Time Window Around Each Transaction",
    fontsize=11
)

ax.set_ylim(0, 125)
ax.grid(axis="y", alpha=0.15)
ax.set_axisbelow(True)

# Remove unnecessary borders
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

fig.text(
    0.5, 0.045,
    "Exact account, currency and amount match | "
    "Matches may occur before or after the labeled payment",
    ha="center",
    fontsize=9,
    color="#475569"
)

fig.text(
    0.5, 0.02,
    "Source: Swift synthetic dataset | "
    "Person 2 verification G7.1 | "
    "N = 679 labeled payments",
    ha="center",
    fontsize=9,
    color="#64748B"
)

plt.tight_layout(rect=[0, 0.09, 1, 1])

plt.savefig(
    "g7_duplicate_repeat_windows.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```
