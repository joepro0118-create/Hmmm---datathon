```python

import matplotlib.pyplot as plt

# Person 2 verified evidence: Rule G7.2
# Comparison uses sender-relative amount ratios

categories = [
    "Unlabeled\nBaseline",
    "Fat-Finger\nLabeled"
]

median_ratios = [1.0, 86.1]

sample_sizes = [1811362, 1843]

# Create figure
fig, ax = plt.subplots(figsize=(10, 6), dpi=150)

colors = ["#94A3B8", "#DC2626"]

bars = ax.bar(
    categories,
    median_ratios,
    color=colors,
    width=0.5
)

# Add labels
for bar, ratio, n in zip(
    bars, median_ratios, sample_sizes
):
    ax.annotate(
        f"{ratio:.1f}x\nN={n:,}",
        xy=(
            bar.get_x() + bar.get_width()/2,
            ratio
        ),
        xytext=(0, 10),
        textcoords="offset points",
        ha="center",
        fontsize=12,
        fontweight="bold"
    )

# Main takeaway
ax.set_title(
    "Fat-Finger Payments Have an 86.1x\n"
    "Median Sender-Relative Amount Ratio",
    fontsize=15,
    fontweight="bold",
    pad=22
)

ax.set_ylabel(
    "Median Transaction-to-Sender Baseline Ratio",
    fontsize=11
)

ax.set_ylim(0, 110)
ax.grid(axis="y", alpha=0.15)
ax.set_axisbelow(True)

# Clean styling
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Scope and boundary
fig.text(
    0.5, 0.05,
    "Only senders with >=5 unlabeled reference payments | "
    "97 labeled rows excluded",
    ha="center",
    fontsize=9,
    color="#475569"
)

fig.text(
    0.5, 0.025,
    "Baseline uses full transaction history, "
    "including future payments",
    ha="center",
    fontsize=9,
    color="#475569"
)

fig.text(
    0.5, 0.005,
    "Source: Swift synthetic dataset | "
    "G7.2",
    ha="center",
    fontsize=8,
    color="#64748B"
)

plt.tight_layout(rect=[0, 0.12, 1, 1])

# Export chart at 300 DPI
plt.savefig(
    "g7_fat_finger_sender_ratio.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```
