```python

import matplotlib.pyplot as plt

# Person 2's verified results
# Source: outputs/rule_tracker.md, Group G3.1
# Population: unlabeled Swift transactions

payment_types = ["C2C", "B2C", "C2B", "B2B"]
median_amounts = [463.65, 3117.51, 3236.32, 22883.00]
sample_sizes = [1512382, 5600965, 609762, 2269885]

colors = ["#93C5FD", "#60A5FA", "#3B82F6", "#1E3A8A"]

# Create figure
fig, ax = plt.subplots(figsize=(11, 6.5), dpi=150)

bars = ax.bar(
    payment_types,
    median_amounts,
    color=colors,
    width=0.60
)

# Add values above bars
for bar, amount, n in zip(bars, median_amounts, sample_sizes):
    ax.annotate(
        f"${amount:,.0f}\nN={n:,}",
        xy=(bar.get_x() + bar.get_width() / 2, amount),
        xytext=(0, 7),
        textcoords="offset points",
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold"
    )

# Titles and labels
ax.set_title(
    "Median B2B Payments Are 49.4x Larger Than C2C Payments",
    fontsize=15,
    fontweight="bold",
    pad=24
)

ax.set_ylabel("Median Transaction Amount (USD)", fontsize=11)
ax.set_xlabel("Payment Type", fontsize=11)

ax.set_ylim(0, 28500)
ax.grid(axis="y", alpha=0.15)
ax.set_axisbelow(True)

# Clean chart appearance
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

fig.text(
    0.5, 0.02,
    "Source: Swift synthetic payments dataset | "
    "Unlabeled transactions | Person 2 verification (G3.1)",
    ha="center",
    fontsize=9,
    color="#666666"
)

plt.tight_layout(rect=[0, 0.06, 1, 1])

# Export high-resolution chart
plt.savefig(
    "g3_party_flow_amounts.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```
