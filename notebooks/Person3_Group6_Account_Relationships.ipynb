```python

import matplotlib.pyplot as plt

# Person 2's verified Group 6.1 results
# Scam labels: Romance, CEO, Account Replacement

observed_overlap = 352
simulation_mean = 401.6

# Central 95% range from 999 simulations
simulation_lower = 386
simulation_upper = 417

categories = ["Observed", "Simulation Mean"]
values = [observed_overlap, simulation_mean]

colors = ["#2563EB", "#94A3B8"]

fig, ax = plt.subplots(figsize=(10, 6), dpi=150)

bars = ax.bar(
    categories,
    values,
    color=colors,
    width=0.5
)

# Add error bar for simulation range
ax.errorbar(
    x=1,
    y=simulation_mean,
    yerr=[
        [simulation_mean - simulation_lower],
        [simulation_upper - simulation_mean]
    ],
    fmt="none",
    ecolor="#1E293B",
    capsize=8,
    linewidth=2,
    label="Central 95% simulation interval"
)

# Add values above bars
for bar, value in zip(bars, values):
    ax.annotate(
        f"{value:.1f}" if value != 352 else "352",
        xy=(
            bar.get_x() + bar.get_width()/2,
            value
        ),
        xytext=(0, 12),
        textcoords="offset points",
        ha="center",
        fontsize=12,
        fontweight="bold"
    )

# Title
ax.set_title(
    "Observed Cross-Scam Beneficiary Overlap\n"
    "Falls Below the Activity-Preserving Benchmark",
    fontsize=14,
    fontweight="bold",
    pad=22
)

ax.set_ylabel(
    "Beneficiaries With Multiple Scam Labels",
    fontsize=11
)

ax.set_ylim(0, 500)
ax.grid(axis="y", alpha=0.15)
ax.set_axisbelow(True)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.legend(loc="upper left", frameon=False)

# Evidence notes
fig.text(
    0.5, 0.045,
    "999 label-shuffle simulations | "
    "Central 95% interval: 386-417",
    ha="center",
    fontsize=9,
    color="#475569"
)

fig.text(
    0.5, 0.02,
    "Source: Swift synthetic dataset | "
    "Person 2 verification G6.1 | Status: Partial",
    ha="center",
    fontsize=9,
    color="#64748B"
)

plt.tight_layout(rect=[0, 0.09, 1, 1])

# Save 300 DPI PNG
plt.savefig(
    "g6_beneficiary_overlap_benchmark.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```
