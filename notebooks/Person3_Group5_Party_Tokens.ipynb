```python

import matplotlib.pyplot as plt

# Verified data from Person 2's token audit
# Group 5: Address-country and BIC8 matching

categories = ["Sender", "Beneficiary"]

matching_records = [8837452, 87555]
eligible_records = [8837452, 8900322]

# Calculate matching percentages
match_rates = [
    matching / eligible * 100
    for matching, eligible in zip(
        matching_records, eligible_records
    )
]

# Create chart
fig, ax = plt.subplots(figsize=(10, 6), dpi=150)

colors = ["#16A34A", "#F59E0B"]

bars = ax.bar(
    categories,
    match_rates,
    color=colors,
    width=0.5
)

# Add labels
for bar, rate, matching, eligible in zip(
    bars,
    match_rates,
    matching_records,
    eligible_records
):
    ax.annotate(
        f"{rate:.2f}%\n"
        f"{matching:,} / {eligible:,}",
        xy=(
            bar.get_x() + bar.get_width()/2,
            rate
        ),
        xytext=(0, 9),
        textcoords="offset points",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

# Headline finding
ax.set_title(
    "Sender Address Countries Match Bank Countries\n"
    "Far More Often Than Beneficiary Addresses",
    fontsize=14,
    fontweight="bold",
    pad=22
)

ax.set_ylabel(
    "Address-Country to BIC8 Match Rate (%)",
    fontsize=11
)

ax.set_ylim(0, 125)
ax.grid(axis="y", alpha=0.15)
ax.set_axisbelow(True)

# Remove unnecessary borders
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Explanation
fig.text(
    0.5, 0.045,
    "Only records with eligible country tokens are included.",
    ha="center",
    fontsize=9,
    color="#475569"
)

fig.text(
    0.5, 0.02,
    "Source: Swift synthetic dataset | "
    "Person 2 token audit, Group 5",
    ha="center",
    fontsize=9,
    color="#64748B"
)

plt.tight_layout(rect=[0, 0.09, 1, 1])

# Save 300 DPI chart
plt.savefig(
    "g5_address_country_asymmetry.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```
