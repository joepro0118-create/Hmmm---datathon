```python

import matplotlib.pyplot as plt

# Person 2's verified findings (Group G1.1)
# Average transactions per UTC day
day_types = ["Weekday", "Weekend"]
avg_transactions = [136603, 42781]
days = [65, 26]

# Calculate percentage reduction
reduction = (
    (avg_transactions[0] - avg_transactions[1])
    / avg_transactions[0]
) * 100

# Create the chart
fig, ax = plt.subplots(figsize=(10, 6), dpi=150)

colors = ["#2563EB", "#93C5FD"]

bars = ax.bar(
    day_types,
    avg_transactions,
    color=colors,
    width=0.5
)

# Add labels
for bar, value, count in zip(bars, avg_transactions, days):
    ax.annotate(
        f"{value:,} tx/day\n({count} days)",
        xy=(bar.get_x() + bar.get_width()/2, value),
        xytext=(0, 8),
        textcoords="offset points",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

# Chart title
ax.set_title(
    f"Weekend Payment Activity Falls by {reduction:.1f}%",
    fontsize=16,
    fontweight="bold",
    pad=22
)

ax.set_ylabel(
    "Average Transactions per UTC Day",
    fontsize=11
)

ax.set_ylim(0, 170000)
ax.grid(axis="y", alpha=0.15)
ax.set_axisbelow(True)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Evidence note
fig.text(
    0.5, 0.02,
    "Source: Swift synthetic dataset | "
    "91 interior UTC dates | Person 2 verification G1.1",
    ha="center",
    fontsize=9,
    color="#666666"
)

plt.tight_layout(rect=[0, 0.06, 1, 1])

# Export chart at 300 DPI
plt.savefig(
    "g1_weekend_rate.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```
