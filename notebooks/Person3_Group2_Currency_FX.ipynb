```python

import matplotlib.pyplot as plt

# Verified Swift data: Person 2, Rule G2.1
# Population: Cross-border payments

categories = [
    "Shared Home\nCurrency",
    "Sender Home\nCurrency Only"
]

# Number of payments with zero FX conversions
zero_fx_count = [547843, 39830]

# Total number of payments in each category
total_count = [797143, 7480937]

# Calculate percentages
zero_fx_rates = [
    count / total * 100
    for count, total in zip(zero_fx_count, total_count)
]

# Create figure
fig, ax = plt.subplots(figsize=(10, 6), dpi=150)

colors = ["#16A34A", "#94A3B8"]

bars = ax.bar(
    categories,
    zero_fx_rates,
    color=colors,
    width=0.5
)

# Add numerical labels
for bar, rate, n in zip(bars, zero_fx_rates, total_count):
    ax.annotate(
        f"{rate:.2f}%\nN={n:,}",
        xy=(bar.get_x() + bar.get_width()/2, rate),
        xytext=(0, 10),
        textcoords="offset points",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

# Chart title
ax.set_title(
    "Shared Home-Currency Payments Have\n"
    "Much Higher Zero-FX Rates",
    fontsize=15,
    fontweight="bold",
    pad=20
)

ax.set_ylabel(
    "Payments With Zero FX Conversions (%)",
    fontsize=11
)

ax.set_ylim(0, 85)
ax.grid(axis="y", alpha=0.15)
ax.set_axisbelow(True)

# Clean chart appearance
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Data source note
fig.text(
    0.5, 0.02,
    "Source: Swift synthetic dataset | "
    "Cross-border transactions | "
    "Person 2 verification G2.1",
    ha="center",
    fontsize=9,
    color="#666666"
)

plt.tight_layout(rect=[0, 0.06, 1, 1])

# Save as 300 DPI PNG
plt.savefig(
    "g2_shared_modal_zero_fx.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```
