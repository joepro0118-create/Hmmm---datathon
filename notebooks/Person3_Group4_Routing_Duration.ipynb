```python

import matplotlib.pyplot as plt

# Person 2 verified results: Rule G4.2
# Scope: 3 participants, continuous paths

routing_types = ["Indirect", "Covered"]

median_duration = [88, 288]
sample_sizes = [1860432, 237474]

colors = ["#60A5FA", "#1E3A8A"]

# Calculate the comparison
difference = median_duration[1] - median_duration[0]
ratio = median_duration[1] / median_duration[0]

# Create chart
fig, ax = plt.subplots(figsize=(10, 6), dpi=150)

bars = ax.bar(
    routing_types,
    median_duration,
    color=colors,
    width=0.5
)

# Display values and sample sizes
for bar, value, n in zip(
    bars, median_duration, sample_sizes
):
    ax.annotate(
        f"{value} sec\nN={n:,}",
        xy=(
            bar.get_x() + bar.get_width()/2,
            value
        ),
        xytext=(0, 9),
        textcoords="offset points",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

# Main finding
ax.set_title(
    f"Covered Payments Have {ratio:.2f}x Higher "
    "Median Duration",
    fontsize=15,
    fontweight="bold",
    pad=22
)

ax.set_ylabel("Median Duration (Seconds)", fontsize=11)

ax.set_ylim(0, 360)
ax.grid(axis="y", alpha=0.15)
ax.set_axisbelow(True)

# Remove unnecessary borders
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Scope annotation
ax.text(
    0.5, 0.90,
    f"+{difference} seconds at 3 participants",
    transform=ax.transAxes,
    ha="center",
    fontsize=11,
    color="#334155"
)

# Evidence notes
fig.text(
    0.5, 0.035,
    "Scope: 3 participants, continuous paths | "
    "Covered has a lower mean and 75th percentile",
    ha="center",
    fontsize=9,
    color="#475569"
)

fig.text(
    0.5, 0.015,
    "Source: Swift synthetic dataset | "
    "Person 2 verification G4.2",
    ha="center",
    fontsize=8,
    color="#64748B"
)

plt.tight_layout(rect=[0, 0.09, 1, 1])

# Export 300 DPI chart
plt.savefig(
    "g4_covered_three_participants.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```
